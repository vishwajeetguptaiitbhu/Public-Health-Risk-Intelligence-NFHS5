## Profiling and Distribution

```python
cd.columns
```

```python
gender_freq = cd['Sex of household member'].value_counts()
print(gender_freq)
```

```python
state_distribution = (
    cd['State']
      .value_counts()
      .to_frame('Count')
)

state_distribution['Percentage'] = (
    state_distribution['Count']
    / len(cd)
    *100
).round(2)

print(state_distribution)
```

```python
wealth_distribution = (
    cd['Wealth index combined']
      .value_counts(normalize=True)
      .mul(100)
      .round(2)
)

print(wealth_distribution)
```

#### CROSS TABULATIION

```python
gender_state = pd.crosstab(
    cd['State'],
    cd['Sex of household member'],
    margins=True
)

print(gender_state)
```

```python
gender_state_percent = pd.crosstab(
    cd['State'],
    cd['Sex of household member'],
    normalize='index'
).round(3) * 100

print(gender_state_percent)
```

```python
gender_residence = pd.crosstab(
    cd['Sex of household member'],
    cd['Type of place of residence'],
    normalize='index'
).round(3)*100

print(gender_residence)
```

```python
residence_wealth = pd.crosstab(
    cd['Type of place of residence'],
    cd['Wealth index combined'],
    normalize='index'
).round(3)*100

print(residence_wealth)
```

```python
state_wealth = pd.crosstab(
    cd['State'],
    cd['Wealth index combined'],
    normalize='index'
).round(3)*100

print(state_wealth)
```

```python
three_way = pd.crosstab(
    [cd['Wealth index combined'],cd['Sex of household member']],
    cd['Type of place of residence'],
    normalize='columns'
).mul(100).round(2)

print(three_way)
```

## CATEGORISATION

```python
def bmi_category(x):

    if x<18.5:
        return "Underweight"

    elif x<25:
        return "Normal"

    elif x<30:
        return "Overweight"

    else:
        return "Obese"

cd["BMI_Category"] = cd["Combined_BMI"].apply(bmi_category)
```

```python
cd.head()
```

```python
cd["Obesity"] = (cd["Combined_BMI"]>=30).astype(int)
```

```python
bins=[15,20,30,40,50,60,100]

labels=[
"15-20",
"21-30",
"31-40",
"41-50",
"51-60",
"60+"
]

cd["Age_Group"] = pd.cut(
cd["Age of household members"],
bins=bins,
labels=labels)
```

```python
cd.to_csv("data/NFHS_FEATURED_DATA.csv")
```
