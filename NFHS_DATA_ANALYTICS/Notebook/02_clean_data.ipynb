```python
import pandas as pd
lst_column = ["HHID", "HVIDX", "HV024", "HV025", "HV026","HV104","HV105","HB2","HA2","HA3","HB3","HA13","HVIDX","HV270","SHDIST","HA40","HB40","BMI"]
# df = pd.read_spss("data/IAPR7EFL.SAV")[lst_column]
import pyreadstat
df, meta = pyreadstat.read_sav(
    "data/IAPR7EFL.SAV",
    usecols=lst_column)
```

```python
df.head()
```

```python
labels=pd.read_csv('data/variables_household_nfhs.csv')
```

```python
labels.head()
```

```python
df.columns
```

```python
for i in df.columns:
    if i in labels['Column_Name'].values:
        print(i)
        if(i=='HB40'):
            df = df.rename(columns={i:"BMI_Man"} )
        elif(i=='HA40'):
            df = df.rename(columns={i:"BMI_Woman"})
        else:
            index=labels[labels['Column_Name']==i].index[0]
            df = df.rename(columns={i: labels['Column_Label'].loc[index]})
```

```python
print(df.columns)
```

```python
df["Combined_BMI"]= df['BMI_Woman'].fillna(df['BMI_Man'])
df["Combined_weight"]= df["Woman's weight in kilograms (1 decimal)"].fillna(df["Man's weight in kilograms (1 decimal)"])
df["Combined_height"]= df["Woman's height in centimeters (1 decimal)"].fillna(df["Man's height in centimeters (1 decimal)"])
df.drop(columns=['BMI_Woman','BMI_Man',"Woman's weight in kilograms (1 decimal)","Man's weight in kilograms (1 decimal)",
                 "Woman's height in centimeters (1 decimal)","Man's height in centimeters (1 decimal)","NA - Place of residence"],inplace= True)
```

```python
df.head(10)
```

```python
df["Unique_ID"]= df["Case Identification"].astype(str) + "_" + df["Line number"].astype(int).astype(str)
```

```python
df.head()
```

```python
df.drop(columns=["Case Identification","Line number","Result of measurement - height/weight"],inplace= True)
```
