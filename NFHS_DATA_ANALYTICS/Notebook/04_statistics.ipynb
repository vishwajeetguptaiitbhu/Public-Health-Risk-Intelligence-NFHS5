## STATISTICAL TESTS

```python
from scipy.stats import chi2_contingency

table = pd.crosstab(df["Sex of household member"],df["Obesity"])
table_percent = pd.crosstab(
    df['Sex of household member'],
    df['Obesity'],
    normalize='index'
) * 100
chi2,p,dof,expected = chi2_contingency(table)
print(table_percent)
print(p)
```

Outcome: A Chi-square test of independence was conducted to examine the association between sex and obesity status. The association was statistically significant (p < 0.001). Among males, 3.65% were obese compared to 5.32% of females, indicating a significantly higher prevalence of obesity among female respondents.

#### Chi-Square Test

```python
urban = df[df['Type of place of residence']==1]

table = pd.crosstab(
    urban['Wealth index combined'],
    urban['Obesity']
)

chi2_contingency(table)
```

```python
rural = df[df['Type of place of residence']==2]

table = pd.crosstab(
    rural['Wealth index combined'],
    rural['Obesity']
)

chi2_contingency(table)
```

Outcome: Chi-square tests revealed a statistically significant association between household wealth and obesity in both urban (χ² = 2100.67, df = 4, p < 0.001) and rural populations (χ² = 10707.25, df = 4, p < 0.001). These findings indicate that obesity prevalence varies significantly across wealth quintiles in both residential settings.

#### LOGISTIC RGRESSION

```python
import statsmodels.formula.api as smf

model = smf.logit(
    "Obesity ~ \
    C(Q('Wealth index combined')) + \
    C(Q('Type of place of residence')) + \
    C(Q('Sex of household member')) + \
    Q('Age of household members')",
    data=df
).fit()

print(model.summary())
```

```python
import numpy as np
import pandas as pd

or_table = pd.DataFrame({
    'Odds Ratio': np.exp(model.params),
    'Lower CI': np.exp(model.conf_int()[0]),
    'Upper CI': np.exp(model.conf_int()[1]),
    'P value': model.pvalues
})

print(or_table)
```

Does wealth affect obesity differently in urban and rural areas?

```python
interaction_model = smf.logit(
    "Obesity ~ \
    C(Q('Wealth index combined')) * C(Q('Type of place of residence')) + \
    C(Q('Sex of household member')) + \
    Q('Age of household members')",
    data=df
).fit()

print(interaction_model.summary()) 
```

```python
odd_table = pd.DataFrame({
    'Odds Ratio': np.exp(interaction_model.params),
    'Lower CI': np.exp(interaction_model.conf_int()[0]),
    'Upper CI': np.exp(interaction_model.conf_int()[1]),
    'P value': interaction_model.pvalues
})

print(odd_table)
```
